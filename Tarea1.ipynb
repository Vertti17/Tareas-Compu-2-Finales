## Explorando un DataFrame 

```python
import pandas as pd
df= pd.read_csv("peliculas.csv")
df.head(1)
```

1. ¿Cuántas películas hay en el dataset?

```python
df.shape[0]
```

Por lo tanto, el dataset contiene 100 películas. 

2. ¿Cuántas variables o columnas tiene el dataset?

```python
df.shape[1]
```

El DataFrame contiene 12 variables o columnas. 

3. ¿Cuáles son las columnas que contiene el DataFrame?

```python
df.columns
```

Al usar `df.columns`podemos ver en forma de lista los nombres de todas las columnas. 

4. ¿Qué tipo de dato tiene cada columna?

```python
df.dtypes
```

Tenemos que `int64` representa números enteros, `float64`números con decimales, `object` representa texto normalmente y `bool`representa valores verdaderos o falsos. 

5. ¿Existen valores faltantes en el dataset?

```python
df.isnull().sum()
```

Podemos ver que tods las columnas tienen 0 valores nulos, por lo tanto, no existen valores faltantes en el dataset. 

6. ¿Cuánta información contiene cada columna? 

```python
df.info()
```

El DataFrame tiene 100 registros, y todas las columnas contienen 100 valores no nulos. Esto confirma que ninguna columna tiene datos faltantes. 

7. ¿Qué columnas contienen información númerica?

```python
df.select_dtypes("number").columns
```

Ahora sabemos que tenemos 7 columnas con información númerica. 

8. ¿Qué columnas contienen información categórica o textual?

```python
df.select_dtypes(["object", "bool"]).columns
```

De igual manera, ahora sabemos que columnas fueron almacenadas como texto, mientras que `adulto` es una variable booleana (`True`o `False`).

9. Obtén únicamente la columna que contiene el título de las películas. 

```python
df["titulo"]
```

Así devuleve solamente la columna de `título`.

10. Obtén únicamente la columna que contiene el rating de las películas. 

```python
df["rating"]
```

Devuelve únicamente el rating de las películas

11. Obtén unicamente las columnas correspondientes al título, género y rating. 

```python
df[["titulo","genero","rating"]]
```

De sta forma obtenemos un nuevo DataFrame solamente con las tres variables solicitadas. 

12. Obtén las estadísticas descriptivas de las variables númericas. 

```python
df.describe()
```

Podemos ver datos como la duración máxima, o los  valores de algunas variables. 

13. ¿Cuál es el rating promedio de las películas?

```python
df["rating"].mean()
```

Esto sucede porque las 100 películas tienen actualmente un rating igual a cero. 

14. ¿Cuál es la duración promedio de las películas?

```python
df["duracion_min"].mean()
```

La duración promedio registrada es de 21.73 minutos, incluyendo las que tienen duración de 0 minutos. 

15. ¿Cuál es la película con mayor duración?

```python
df.groupby("titulo")[["duracion_min"]].max().sort_values(by="duracion_min",ascending=False).head(1)
```

La película con mayor duración es American Static , con 186 minutos. 

16. ¿Cuál es la película con mayor rating?

```python
df.groupby("titulo")[["rating"]].max().sort_values(by="rating",ascending=False).head(1)
```

Todas las películas estan empatadas. Devuelve `100 Years` porque es la primera película del DataFrame, no porque tenga un rating mayor que las demás. 

17. ¿Cuál es la película que recibió mas votos?

```python
df.groupby("titulo")[["votos"]].max().sort_values(by="votos",ascending=False).head(1)
```

Al igual que con el rating, todas las películas tienen 0 votos, por lo que estan empatadas y devuelve el primer registro. 

18. ¿Qué géneros aparecen en el dataset y cuántas películas hay de cada género?

```python
df.groupby("genero")[["titulo"]].count().sort_values(by="titulo",ascending=False)
```

Aquí tenemos los géneros que hay en el dataset y de mayor a menor cuántas películas hay por cada género. 

19. ¿Qué país aparece con mayor frecuencia?
El dataset proporcionado no contiene ninguna columna llamada `pais` ni otra columna que indique directamente el país de cada película. Sin embargo, el código quedaría de esta manera:

```python
df.groupby("pais")[["titulo"]].count().sort_values(by="titulo", ascending=False).head(1)
```

20. ¿Qué plataforma tiene más películas? Sucede lo mismo con el caso anterior. El dataset no contiene ninguna variable llamada `plataforma`. Pero de igual manera, el código se vería así:

```python
df.groupby("plataforma")[["titulo"]].count().sort_values(by="titulo", ascending=False).head(1)
```
