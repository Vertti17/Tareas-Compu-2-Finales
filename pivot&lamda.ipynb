```python
import pandas as pd 
import numpy as np
orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
order_reviews = pd.read_csv("olist_order_reviews_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
geo = pd.read_csv("olist_geolocation_dataset.csv")
order_payments = pd.read_csv("olist_order_payments_dataset.csv")
```

```python
df1=orders.merge(order_items,on="order_id",how="left")
df1=df1.merge(products,on="product_id",how="left")
```

```python
# 9. Utilizar fill_value=0 y comparar con la tabla original. ¿Qué combinaciones no tenían observaciones?
p9 = df1.pivot_table(
    values="price",
    index="product_category_name",
    columns="order_status",
    aggfunc="sum",
    fill_value=0
)
display(p9.head())
```

Interpretación Ejercicio 9:
Al usar fill_value=0, los valores NaN de la tabla original (que significan que NO hubo observaciones para esa combinación de categoría y estado) se reemplazan por 0. 
Por ejemplo, en la tabla original, la categoría 'agro_industria_e_comercio' tiene NaN en 'approved', 'canceled', 'processing', etc. 
Esto significa que no hubo pedidos en esos estados para esa categoría. Con fill_value=0, ahora se muestran como 0, indicando ausencia de ventas.

```python
# 10. Pivot table utilizando más de una variable en index. Interpretación.
p10 = df1.pivot_table(
    values="price",
    index=["product_category_name", "order_status"],
    aggfunc="mean"
)
display(p10.head(10))
```

Interpretación Ejercicio 10:
El índice tiene dos niveles (MultiIndex). El primer nivel representa la categoría del producto (product_category_name) y el segundo nivel representa el estado del pedido (order_status). 
Los valores mostrados son el precio promedio para esa combinación específica. Por ejemplo, para 'agro_industria_e_comercio', el precio promedio de los pedidos 'delivered' es aproximadamente 141.13.

```python
# 11. Crear columna 'total' que sea la suma de 'price' y 'freight_value'
# Nota: Usamos apply con lambda para cumplir con el requisito del nivel, aunque df1['price'] + df1['freight_value'] es más eficiente.
df1['total'] = df1.apply(lambda row: row['price'] + row['freight_value'] if pd.notnull(row['price']) and pd.notnull(row['freight_value']) else np.nan, axis=1)
display(df1[['price', 'freight_value', 'total']].head())
```

```python
# 12. Crear columna que clasifique los productos como "barato" o "caro" según un umbral.
# Justificación del umbral: Usaremos la mediana del precio (aprox 74.99) como umbral. 
# La mediana es robusta ante valores atípicos (outliers), a diferencia del promedio.
umbral_precio = df1['price'].median()
print(f"\nUmbral seleccionado (Mediana): {umbral_precio:.2f}")

df1['clasificacion_precio'] = df1['price'].apply(lambda x: "barato" if pd.notnull(x) and x <= umbral_precio else "caro")

display(df1[['price', 'clasificacion_precio']].head())
```

```python
# 13. Columna que indique si el costo de envío representa más del 20% del precio del producto.
df1['envio_mayor_20_pct'] = df1.apply(lambda row: True if pd.notnull(row['price']) and row['price'] > 0 and (row['freight_value'] / row['price']) > 0.20 else False, axis=1)
display(df1[['price', 'freight_value', 'envio_mayor_20_pct']].head())
```

```python
# 14. Nueva columna que convierta el nombre de la categoría a mayúsculas.
df1['categoria_mayusculas'] = df1['product_category_name'].apply(lambda x: str(x).upper())
display(df1[['product_category_name', 'categoria_mayusculas']].head())
```

```python
# 15. Columna que indique si el nombre de la categoría contiene una palabra determinada.
palabra_clave = "casa" # Ejemplo de palabra clave
df1['contiene_palabra'] = df1['product_category_name'].apply(lambda x: palabra_clave in str(x).lower())
display(df1[['product_category_name', 'contiene_palabra']].head())
```

```python
# 16. Clasificar pedidos según el costo total y construir pivot table con el costo total promedio por categoría y segmento.
# Clasificaremos usando cuantiles para crear 3 segmentos: Bajo, Medio, Alto
df1['segmento_costo'] = pd.qcut(df1['total'], q=3, labels=['Bajo', 'Medio', 'Alto'])

p16 = df1.pivot_table(
    values='total',
    index='product_category_name',
    columns='segmento_costo',
    aggfunc='mean'
)
display(p16.head())
```

```python
# 17. Calcular porcentaje de envío vs precio, clasificar ("envío bajo", "envío alto") y comparar mediante pivot table.
# Clasificaremos como "envío alto" si supera el 20% del precio.
df1['porcentaje_envio'] = df1.apply(lambda row: (row['freight_value'] / row['price']) * 100 if pd.notnull(row['price']) and row['price'] > 0 else 0, axis=1)
df1['clasificacion_envio'] = df1['porcentaje_envio'].apply(lambda x: 'envío alto' if x > 20 else 'envío bajo')

p17 = df1.pivot_table(
    values='price',
    index='clasificacion_envio',
    columns='order_status',
    aggfunc='mean'
)
display(p17)
```

```python
# 18. Variable que indique si el pedido tiene un precio superior al precio promedio de todos los productos. Analizar con pivot table.
precio_promedio_global = df1['price'].mean()
df1['precio_superior_promedio'] = df1['price'].apply(lambda x: 'Superior' if pd.notnull(x) and x > precio_promedio_global else 'Inferior')

p18 = df1.pivot_table(
    values='total',
    index='precio_superior_promedio',
    columns='order_status',
    aggfunc='count'
)
display(p18)
```

```python
# 19. Pivot table que muestre, para cada categoría, el total y el promedio de una variable numérica relevante (usaremos 'total').
p19 = df1.pivot_table(
    values='total',
    index='product_category_name',
    aggfunc=['sum', 'mean']
)
display(p19.head())
```

```python
# 20. Pregunta de negocio: ¿Cuál es el ingreso total generado por los productos clasificados como 'caro' en comparación con los 'barato', desglosado por estado del pedido?

print("Pregunta: ¿Cuál es el ingreso total (price) generado por productos 'caro' vs 'barato' según el estado del pedido?")

# Transformación con Lambda: Ya tenemos 'clasificacion_precio' del ejercicio 12.
# Aplicamos pivot_table
p20 = df1.pivot_table(
    values='price',
    index='clasificacion_precio',
    columns='order_status',
    aggfunc='sum',
    fill_value=0
)
print("\nPivot Table Resultante:")
display(p20)

print("""
Interpretación: 
Los productos 'caro' (por encima de la mediana) generan significativamente más ingresos en todos los estados de pedido. 
La mayor contribución se da en pedidos 'delivered' (entregados), donde los productos caros suman más de 13 millones, frente a poco más de 2 millones de los baratos.
""")
```

```python
# 21. Relación entre precio, costo de envío y categoría. Crear variable derivada con Lambda y usar pivot table.
# Variable derivada: ratio de envío (freight/price)
df1['ratio_envio'] = df1.apply(lambda row: row['freight_value'] / row['price'] if pd.notnull(row['price']) and row['price'] > 0 else 0, axis=1)

p21 = df1.pivot_table(
    values=['price', 'freight_value', 'ratio_envio'],
    index='product_category_name',
    aggfunc='mean'
)
print("Pivot table de promedios por categoría:")
display(p21.head(10))
```

```python
# 22. Reporte: ¿Qué categorías presentan comportamientos diferentes dependiendo del estado del pedido?
print("\n--- Ejercicio 22: Reporte de Comportamiento por Estado ---")
# Pivot 1: Conteo de pedidos por categoría y estado
p22_1 = df1.pivot_table(
    values='order_id',
    index='product_category_name',
    columns='order_status',
    aggfunc='count',
    fill_value=0
)

# Pivot 2: Tasa de cancelación (Canceled / Total) por categoría
# Transformación Lambda: Calcular tasa de cancelación por fila
p22_2 = p22_1.apply(lambda row: (row['canceled'] / row.sum()) * 100 if row.sum() > 0 else 0, axis=1).to_frame(name='tasa_cancelacion_%')

print("Pivot 1: Conteo de pedidos")
display(p22_1.head())
print("\nPivot 2: Tasa de cancelación (%)")
display(p22_2.sort_values(by='tasa_cancelacion_%', ascending=False).head(10))

print("""
Interpretación:
Categorías como 'audio' o 'telefonia' pueden presentar tasas de cancelación notables. 
Al cruzar el estado del pedido con la categoría, se observa que la mayoría de los pedidos se concentran en 'delivered', pero existen nichos con alta volatilidad (cancelaciones o no disponibles).
""")
```

```python
# RETO: Preparar un resumen para la dirección usando lambda, apply(), pivot_table() y aggfunc.

print("--- RETO FINAL: Resumen Ejecutivo para la Dirección ---")

# 1. Transformación con Lambda y apply():
# Clasificar el costo de envío como 'Aceptable' o 'Excesivo' (si supera el 30% del precio)
df1['clasificacion_costo_envio'] = df1.apply(
    lambda row: 'Excesivo' if pd.notnull(row['price']) and row['price'] > 0 and (row['freight_value'] / row['price']) > 0.30 else 'Aceptable', 
    axis=1
)

# 2. Pivot Table con aggfunc múltiple para el resumen ejecutivo
resumen_direccion = df1.pivot_table(
    values=['price', 'freight_value', 'total'],
    index='product_category_name',
    columns='clasificacion_costo_envio',
    aggfunc={
        'price': 'mean',
        'freight_value': 'mean',
        'total': 'sum'
    },
    fill_value=0
)

print("Resumen de métricas por Categoría y Clasificación de Costo de Envío:")
display(resumen_direccion.head(15))

# 3. Añadir un análisis de volumen de pedidos usando aggfunc='count'
volumen_pedidos = df1.pivot_table(
    values='order_id',
    index='product_category_name',
    columns='order_status',
    aggfunc='count',
    fill_value=0
)

print("\nVolumen de pedidos por categoría y estado (Top 10):")
display(volumen_pedidos.head(10))

print("""
CONCLUSIÓN DEL RETO:
Este resumen permite a la dirección identificar qué categorías tienen costos de envío 'Excesivos' (superiores al 30% del precio). 
Aquellas categorías con altos promedios de 'total' y altos volúmenes en 'delivered' son las más rentables. 
Se recomienda revisar las categorías que concentran una gran cantidad de pedidos en 'canceled' o 'unavailable', así como aquellas donde el costo de envío 'Excesivo' es la norma, para optimizar la logística o ajustar los precios.
""")
```
