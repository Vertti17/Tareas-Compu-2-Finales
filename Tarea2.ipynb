# 1. Importar Pandas

Pandas es una librería de Python que permite trabajar y analizar datos, principalmente mediante estructuras como los DataFrames.

Para utilizar Pandas primero debemos importar la librería. El alias que normalmente se utiliza es `pd`.

**Código principal:** `import pandas as pd`


```python
import pandas as pd
```

### Ejemplo con nuestra base de datos

Después de importar Pandas podemos cargar nuestra base de datos y guardarla en un DataFrame llamado `df`.

El resultado es que Pandas queda disponible para utilizar sus funciones y métodos durante todo el análisis.


```python
df = pd.read_csv("base_generica_pandas.csv")
df
```

# 2. Cargar una base de datos

Para cargar un archivo CSV utilizamos la función `read_csv()` de Pandas.

La sintaxis general es `pd.read_csv("archivo.csv")`.

**Código principal:** `pd.read_csv("archivo.csv")`


```python
df = pd.read_csv("base_generica_pandas.csv")
```

### Ejemplo con nuestra base de datos

Nuestro archivo se llama `base_generica_pandas.csv`, por lo que utilizamos `pd.read_csv()` para leerlo y guardarlo en `df`.

**Código principal:** `df.head()`

Muestra las últimas filas del DataFrame. Por defecto muestra 5 filas.

El resultado muestra los últimos 8 registros de nuestra base.

El atributo `shape` muestra el tamaño del DataFrame. El primer número corresponde a las filas y el segundo a las columnas.

**Código principal:** `df.shape`


```python
df.shape
```

### Resultado

### `columns`

El atributo `columns` permite conocer los nombres de todas las columnas del DataFrame.

**Código principal:** `df.columns`


```python
df.columns
```

### Resultado

### `info()`

El método `info()` proporciona información general sobre las columnas, los valores no nulos y los tipos de datos.

**Código principal:** `df.info()`


```python
df.info()
```

### Resultado

El resultado permite conocer la estructura general de las 100 entradas y 10 columnas de nuestra base.


# 4. Seleccionar columnas

Pandas permite seleccionar una o varias columnas utilizando corchetes `[]`.

Para una columna usamos `df["columna"]`. Para varias columnas usamos una lista dentro de los corchetes, por ejemplo `df[["col1", "col2"]]`.


### Una columna

Para seleccionar una sola columna escribimos su nombre entre comillas.

**Código principal:** `df["col5"]`


```python
df["col5"]
```

### Resultado

El resultado es una Serie de Pandas que contiene todos los valores de `col5`.


### Varias columnas

Para seleccionar varias columnas utilizamos una lista.

**Código principal:** `df[["col1", "col2"]]`


```python
df[["col1", "col2"]]
```

### Resultado

El resultado es un DataFrame que contiene únicamente las columnas seleccionadas.


### Selección mediante listas

Podemos guardar los nombres de las columnas que queremos seleccionar en una lista y utilizarla después.

**Código principal:** `columnas_seleccionadas = ["col1", "col5", "col10"]`


```python
columnas_seleccionadas = ["col1", "col5", "col10"]
df[columnas_seleccionadas]
```

### Resultado

El resultado contiene únicamente `col1`, `col5` y `col10`.


# 5. Filtrar datos

Filtrar datos significa seleccionar únicamente las filas que cumplen una condición.

Podemos utilizar operadores como `>`, `<`, `>=`, `<=`, `==` y `!=`.

**Código principal:** `df[df["col1"] > 500]`


### Condiciones numéricas

Podemos filtrar una columna numérica mediante una condición.


```python
df[df["col1"] > 500]
```

### Resultado

El resultado contiene únicamente las filas donde `col1` es mayor que 500.


### Condiciones categóricas

También podemos filtrar utilizando variables categóricas.

Por ejemplo, podemos seleccionar los registros donde `col6` sea `"Sur"`.


```python
df[df["col6"] == "Sur"]
```

### Resultado

El resultado muestra únicamente los registros cuya categoría en `col6` es `"Sur"`.


### Varias condiciones

Podemos combinar condiciones utilizando:
- `&` para AND.
- `|` para OR.
- `~` para NOT.

Cuando combinamos condiciones debemos colocar cada una entre paréntesis.


```python
df[(df["col6"] == "Sur") & (df["col5"] != "B")]
```

### Resultado

El resultado muestra las filas que cumplen simultáneamente que `col6` sea `"Sur"` y `col5` sea diferente de `"B"`.

También podemos utilizar OR:


```python
df[(df["col5"] == "A") | (df["col6"] == "Norte")]
```

### Resultado

El resultado incluye los registros donde se cumple al menos una de las dos condiciones.


# 6. Ordenar datos

El método `sort_values()` permite ordenar las filas utilizando una o varias columnas.

Por defecto ordena de menor a mayor. Para ordenar de mayor a menor utilizamos `ascending=False`.

**Código principal:** `df.sort_values("col1")`


```python
df.sort_values("col1")
```

### Resultado

Los registros quedan ordenados de menor a mayor según `col1`.


### Ejemplo con nuestra base de datos

Podemos ordenar `col1` de mayor a menor.


```python
df.sort_values("col1", ascending=False)
```

### Resultado

Los registros con los valores más altos de `col1` aparecen primero.

También podemos ordenar utilizando varias columnas.


```python
df.sort_values(["col10", "col1"])
```

### Resultado

Primero se ordena por `col10` y después por `col1` dentro de cada categoría.


# 7. Estadística descriptiva

Las estadísticas descriptivas permiten resumir las variables numéricas de nuestra base.

- `describe()` = resumen estadístico.
- `mean()` = promedio.
- `median()` = mediana.
- `min()` = mínimo.
- `max()` = máximo.
- `std()` = desviación estándar.


### `describe()`

Genera un resumen estadístico de las columnas numéricas.

**Código principal:** `df.describe()`


```python
df.describe()
```

### Resultado

El resultado incluye cantidad de datos, media, desviación estándar, mínimo, cuartiles y máximo de las variables numéricas.


### `mean()`

Calcula la media o promedio de una columna.

**Código principal:** `df["col1"].mean()`


```python
df["col1"].mean()
```

### Resultado

La media de `col1` se calcula a partir de los 100 registros de nuestra base.


### `median()`

Calcula la mediana de una columna, es decir, el valor central de los datos ordenados.

**Código principal:** `df["col3"].median()`


```python
df["col3"].median()
```

### Resultado

El resultado es la mediana de `col3`.


### `min()`

Encuentra el valor mínimo de una columna.

**Código principal:** `df["col1"].min()`


```python
df["col1"].min()
```

### Resultado

El resultado es el menor valor registrado en `col1`.


### `max()`

Encuentra el valor máximo de una columna.

**Código principal:** `df["col1"].max()`


```python
df["col1"].max()
```

### Resultado

El resultado es el mayor valor registrado en `col1`.


### `std()`

Calcula la desviación estándar de una columna y permite conocer la dispersión de los datos respecto a su media.

**Código principal:** `df["col1"].std()`


```python
df["col1"].std()
```

### Resultado

El resultado muestra la desviación estándar de `col1`.


# 8. Trabajar con variables categóricas

Las variables categóricas contienen diferentes categorías o grupos.

- `value_counts()` = cuenta la frecuencia de cada categoría.
- `unique()` = muestra las categorías diferentes.
- `nunique()` = cuenta cuántas categorías diferentes existen.


### `value_counts()`

Cuenta cuántas veces aparece cada categoría en una columna.

**Código principal:** `df["col5"].value_counts()`


```python
df["col5"].value_counts()
```

### Resultado

En nuestra base, `col5` contiene las categorías A, B, C y D y el resultado muestra cuántas veces aparece cada una.


### `unique()`

Muestra los diferentes valores que aparecen en una columna sin repetirlos.

**Código principal:** `df["col6"].unique()`


```python
df["col6"].unique()
```

### Resultado

En `col6` podemos observar las diferentes categorías presentes en la base.


### `nunique()`

Cuenta cuántos valores diferentes existen en una columna.

**Código principal:** `df["col6"].nunique()`


```python
df["col6"].nunique()
```

### Resultado

El resultado indica cuántas categorías diferentes existen en `col6`.


# 9. Agrupar datos

El método `groupby()` permite dividir los datos en grupos utilizando una variable categórica y posteriormente calcular estadísticas para cada grupo.

Esto es especialmente útil para responder preguntas como:

- ¿Cuál es el promedio de `col1` para cada categoría de `col5`?
- ¿Cuántos registros existen en cada categoría?
- ¿Cuál es el valor máximo y mínimo de cada grupo?

**Código principal:** `df.groupby("col5")`


### Promedios por categoría

Podemos utilizar `groupby()` junto con `mean()` para calcular el promedio de una variable numérica para cada categoría.


```python
df.groupby("col5")["col1"].mean()
```

### Resultado

El resultado muestra el promedio de `col1` para cada categoría de `col5`.


### Conteos por categoría

También podemos contar cuántos registros existen en cada categoría utilizando `size()`.


```python
df.groupby("col6").size()
```

### Resultado

El resultado muestra la cantidad de registros que pertenece a cada categoría de `col6`.


### Agregaciones

Con `agg()` podemos calcular varias estadísticas al mismo tiempo.

En este ejemplo obtenemos el promedio, máximo, mínimo y cantidad de registros de `col1` para cada categoría de `col10`.


```python
df.groupby("col10")["col1"].agg(["mean", "max", "min", "count"])
```

### Resultado

El resultado es una tabla donde cada fila representa una categoría de `col10` y las columnas contienen diferentes estadísticas de `col1`.


# 10. Crear y modificar columnas

Podemos crear nuevas columnas a partir de las variables existentes.

La sintaxis general es:

`df["nueva_columna"] = ...`


### Ejemplo con nuestra base de datos

Crearemos una nueva columna llamada `col1_mas_col3` sumando los valores de `col1` y `col3`.


```python
df["col1_mas_col3"] = df["col1"] + df["col3"]
df[["col1", "col3", "col1_mas_col3"]].head()
```

### Resultado

La nueva columna contiene, para cada fila, la suma de `col1` y `col3`.


# 11. Extra: combinar operaciones

Una de las ventajas de Pandas es que podemos combinar varias operaciones para responder preguntas más específicas.

En lugar de utilizar los comandos de manera aislada, podemos construir expresiones que primero filtren, después seleccionen y finalmente ordenen los datos.

**Objetivo:** dejar de pensar en comandos aislados y comenzar a construir soluciones utilizando varias operaciones de Pandas.


```python
df[df["col1"] > 500][["col1", "col5", "col6", "col10"]].sort_values("col1", ascending=False)
```

### Resultado

En esta expresión:
1. Filtramos los registros donde `col1` es mayor que 500.
2. Seleccionamos `col1`, `col5`, `col6` y `col10`.
3. Ordenamos los resultados de mayor a menor según `col1`.

El resultado es una tabla con los registros que cumplen la condición, mostrando únicamente las columnas seleccionadas.


### Ejemplo final

También podemos combinar filtrado y agrupación para responder una pregunta más específica: ¿cuál es el promedio de `col1` para cada categoría de `col5`, considerando únicamente los registros donde `col1` es mayor que 500?


```python
df[df["col1"] > 500].groupby("col5")["col1"].mean()
```

### Resultado

La expresión primero filtra los registros con `col1 > 500` y después calcula el promedio de `col1` para cada categoría de `col5`.

Esto demuestra cómo podemos construir soluciones combinando diferentes operaciones de Pandas.
